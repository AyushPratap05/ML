```python
# Display all DataFrame rows where 'Duration_hour' is not a number
df[pd.to_numeric(df['Duration_hour'], errors='coerce').isna()]
```

```python
mask = df['Duration_hour'] == '5m'

df.loc[mask, 'Duration_min'] = 5

df.loc[mask, 'Duration_hour'] = 0

# 4. Now safely convert both columns to int (filling any remaining NaNs with 0)
df['Duration_hour'] = df['Duration_hour'].fillna(0).astype(int)
df['Duration_min'] = df['Duration_min'].fillna(0).astype(int)
```
