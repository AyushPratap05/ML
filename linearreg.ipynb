```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# x_train is the input variable (size in 1000 sq feet)
# y_train is the target (price in 1000) dollars

x_train = np.array([1.0,2.0])
y_train = np.array([300.0,500.0])

print(f"x_train={x_train}")
print(f"y_train={y_train}")
```

```python
plt.scatter(x_train,y_train,marker='x',c='r')

plt.title("Housing price")

plt.ylabel('price (in 1000s of dollars)')
plt.xlabel('size(1000 sqft)')
plt.show()
```

```python
# MODEL FUNCTION f(x) = WX^I + B 
w = 200 
b = 100 
print(f"w:{w},b:{b}")
```

```python
def compute_model_output(x,w,b):

     m = x_train.shape[0]
     f_wb = np.zeros(m)
     for i in range(m):
          f_wb[i] = w * x[i] + b
     return f_wb
```

```python
tmp_f_wb = compute_model_output(x_train, w, b,)

# Plot our model prediction
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')

# Plot the data points
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')

# Set the title
plt.title("Housing Prices")
# Set the y-axis label
plt.ylabel('Price (in 1000s of dollars)')
# Set the x-axis label
plt.xlabel('Size (1000 sqft)')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Generate Synthetic Data (y = 2x + 1)
np.random.seed(42)
X = np.array([1, 2, 3, 4, 5])
y = np.array([3, 5, 7, 9, 11])

# 2. Compute Cost J(w, b) across a grid of w and b values
w_vals = np.linspace(-1, 5, 100)
b_vals = np.linspace(-3, 5, 100)
W, B = np.meshgrid(w_vals, b_vals)
J = np.zeros(W.shape)

m = len(X)
for i in range(W.shape[0]):
    for j in range(W.shape[1]):
        w_tmp = W[i, j]
        b_tmp = B[i, j]
        predictions = w_tmp * X + b_tmp
        J[i, j] = (1 / (2 * m)) * np.sum((predictions - y) ** 2)

# 3. Create Plots
fig = plt.figure(figsize=(14, 6))

# Panel 1: 3D Surface Plot
ax1 = fig.add_subplot(1, 2, 1, projection='3d')
surf = ax1.plot_surface(W, B, J, cmap='viridis', alpha=0.85)
ax1.set_xlabel('Weight (w)')
ax1.set_ylabel('Bias (b)')
ax1.set_zlabel('Cost J(w,b)')
ax1.set_title('3D Cost Surface (Bowl Shape)')
ax1.scatter([2], [1], [0], color='red', s=50, label='Global Minimum (2, 1)')

# Panel 2: 2D Contour Plot
ax2 = fig.add_subplot(1, 2, 2)
contour = ax2.contour(W, B, J, levels=np.logspace(-1, 3, 20), cmap='viridis')
ax2.clabel(contour, inline=True, fontsize=8)
ax2.plot(2, 1, 'r*', markersize=12, label='Optimal (w=2, b=1)')
ax2.set_xlabel('Weight (w)')
ax2.set_ylabel('Bias (b)')
ax2.set_title('Contour Map (Top-Down View)')
ax2.grid(True)
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from sklearn.metrics import mean_squared_error, mean_absolute_error

# Actual values vs Model Predictions
y_actual = np.array([3.0, 5.0, 7.0, 9.0])
y_pred   = np.array([2.5, 5.5, 6.0, 10.0])

# Calculations
mae = mean_absolute_error(y_actual, y_pred)
mse = mean_squared_error(y_actual, y_pred)
rmse = np.sqrt(mse)

print(f"MAE : {mae:.2f}")  # Average absolute error
print(f"MSE : {mse:.2f}")  # Squared error (cost)
print(f"RMSE: {rmse:.2f}") # Error in target units
```
