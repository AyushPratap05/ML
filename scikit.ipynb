```python
from sklearn.linear_model import LogisticRegression

# Data
X = [[0, 0], [1, 1], [2, 2], [3, 3]]  # Features
y = [0, 0, 1, 1]                       # Labels

# Create estimator
model = LogisticRegression()

# Fit (learn from data)
model.fit(X, y)

# Predict
predictions = model.predict([[1.5, 1.5], [2.5, 2.5]])
print(predictions)  # [0, 1]
```
